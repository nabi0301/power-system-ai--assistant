````python
# Smart AI Assistant Functions
def detect_user_intent(message, context):
    """Advanced intent detection using pattern recognition and context"""
    message_lower = message.lower()
    
    # Advanced intent patterns for smarter understanding
    intent_patterns = {
        'exploration': ['explore', 'discover', 'find', 'search', 'investigate', 'look for'],
        'comparison': ['compare', 'versus', 'vs', 'difference', 'contrast', 'against'],
        'optimization': ['optimize', 'improve', 'enhance', 'better', 'efficient', 'best'],
        'prediction': ['predict', 'forecast', 'trend', 'future', 'what if', 'scenario'],
        'explanation': ['why', 'how', 'explain', 'reason', 'cause', 'tell me'],
        'troubleshooting': ['problem', 'issue', 'error', 'wrong', 'fix', 'solve'],
        'learning': ['learn', 'understand', 'teach', 'show me', 'tutorial', 'guide'],
        'analysis': ['analyze', 'analysis', 'examine', 'study', 'evaluate']
    }
    
    detected_intents = []
    for intent, keywords in intent_patterns.items():
        if any(keyword in message_lower for keyword in keywords):
            detected_intents.append(intent)
    
    # Add to context for learning
    if detected_intents:
        context['analysis_patterns']['last_intent'] = detected_intents[0]
    
    return detected_intents

def generate_smart_suggestions(current_context, analysis_type='general'):
    """Generate intelligent suggestions based on context and patterns"""
    suggestions = []
    
    base_suggestions = {
        'voltage': [
            "💡 Try 'smart analysis' for AI-powered voltage insights",
            "🔍 Ask for 'pattern analysis' to detect voltage anomalies", 
            "📊 Request 'predictive analysis' for voltage forecasting",
            "⚡ Explore 'voltage optimization recommendations'"
        ],
        'loading': [
            "⚡ Try 'smart analysis' for intelligent loading assessment",
            "🔄 Ask for 'pattern analysis' to identify loading patterns",
            "📈 Request 'predictive analysis' for capacity planning",
            "🎯 Explore 'thermal limit optimization'"
        ],
        'comparison': [
            "📊 Try 'smart analysis' for SLR vs DLR intelligence",
            "💰 Ask for 'optimization analysis' for efficiency improvements",
            "🎯 Request 'predictive scenarios' for future planning",
            "🔬 Explore 'cost-benefit analysis'"
        ],
        'general': [
            "🧠 Try 'smart analysis' for AI-powered insights",
            "🔍 Ask for 'pattern analysis' to detect anomalies",
            "🔮 Request 'predictive analysis' for forecasting",
            "💡 Explore 'intelligent recommendations'"
        ]
    }
    
    suggestions = base_suggestions.get(analysis_type, base_suggestions['general'])
    
    # Add contextual suggestions based on conversation history
    if len(current_context['conversation_history']) > 0:
        last_topics = [msg.get('message', '') for msg in current_context['conversation_history'][-3:]]
        if any('violation' in topic.lower() for topic in last_topics):
            suggestions.append("🔧 Consider asking about 'mitigation strategies'")
        if any('trend' in topic.lower() for topic in last_topics):
            suggestions.append("📈 Ask about 'long-term forecasting'")
    
    return suggestions[:4]  # Return top 4 suggestions

def perform_smart_analysis(data_type, data=None, context=None):
    """Perform intelligent analysis with AI-powered insights"""
    insights = []
    
    try:
        if data_type == 'voltage' and data is not None:
            # AI-powered voltage analysis
            voltage_mean = data['VM'].mean()
            voltage_std = data['VM'].std()
            voltage_range = data['VM'].max() - data['VM'].min()
            
            # AI pattern detection
            if voltage_std > 0.05:
                insights.append("🚨 **AI Alert:** High voltage variability detected - potential grid instability")
            if voltage_mean < 0.98:
                insights.append("⚠️ **AI Insight:** System-wide voltage depression - consider voltage support")
            if voltage_range > 0.15:
                insights.append("📊 **AI Pattern:** Wide voltage range suggests uneven load distribution")
            
            # Smart recommendations
            low_voltage_buses = data[data['VM'] < 0.95]
            if len(low_voltage_buses) > 0:
                insights.append(f"🎯 **AI Recommendation:** {len(low_voltage_buses)} buses need voltage support")
            else:
                insights.append("✅ **AI Assessment:** Voltage levels within acceptable ranges")
                
        elif data_type == 'loading' and data is not None:
            # AI-powered loading analysis
            valid_data = data[data['RATE'] > 0]
            loading_pct = (valid_data['MVA'] / valid_data['RATE'] * 100)
            
            high_loading = loading_pct[loading_pct > 85]
            critical_loading = loading_pct[loading_pct > 95]
            
            if len(high_loading) > len(loading_pct) * 0.2:
                insights.append("🔥 **AI Alert:** 20%+ lines heavily loaded - system stress detected")
            
            if len(critical_loading) > 0:
                insights.append(f"⚡ **AI Prediction:** {len(critical_loading)} lines at risk of overload")
            else:
                insights.append("✅ **AI Assessment:** No immediate overload risk detected")
                
            # Predictive insights
            avg_loading = loading_pct.mean()
            if avg_loading > 70:
                insights.append("📈 **AI Forecast:** Consider capacity expansion planning")
        
        else:
            # General AI insights
            insights.append("🧠 **AI Analysis:** Ready to provide intelligent insights with data")
            insights.append("💡 **Smart Tip:** Upload power system data for advanced AI analysis")
        
        # Build response
        response = "🧠 **AI-Powered Smart Analysis:**\n\n"
        if insights:
            response += "\n".join(insights)
        else:
            response += "✅ **AI Assessment:** System parameters within normal ranges"
        
        response += "\n\n🎯 **AI Recommendation:** Continue monitoring for pattern changes"
        
        # Add to context
        if context:
            context['recent_insights'].append({
                'timestamp': datetime.now(),
                'type': 'smart_analysis',
                'insights': insights
            })
        
        return response
        
    except Exception as e:
        return f"❌ **AI Error:** Unable to perform smart analysis: {str(e)}"

def perform_pattern_analysis(data_type, data=None, context=None):
    """AI-powered pattern recognition and anomaly detection"""
    try:
        response = "🔍 **AI Pattern Analysis:**\n\n"
        
        if data_type == 'voltage' and data is not None:
            # Statistical pattern analysis
            skewness = data['VM'].skew()
            kurtosis = data['VM'].kurtosis()
            
            response += "📊 **Voltage Patterns Detected:**\n"
            
            if abs(skewness) > 0.5:
                response += f"• Asymmetric distribution detected (skewness: {skewness:.2f})\n"
            else:
                response += "• Normal distribution pattern confirmed\n"
                
            if kurtosis > 3:
                response += "• High peak concentration - some buses have similar voltages\n"
            elif kurtosis < -1:
                response += "• Flat distribution - wide voltage spread across system\n"
            else:
                response += "• Normal voltage distribution pattern\n"
                
            # Anomaly detection
            q1, q3 = data['VM'].quantile([0.25, 0.75])
            iqr = q3 - q1
            outliers = data[(data['VM'] < q1 - 1.5*iqr) | (data['VM'] > q3 + 1.5*iqr)]
            
            if len(outliers) > 0:
                response += f"• {len(outliers)} voltage outliers detected - investigate these buses\n"
            else:
                response += "• No voltage outliers detected\n"
                
        elif data_type == 'loading' and data is not None:
            response += "📈 **Loading Patterns Detected:**\n"
            valid_data = data[data['RATE'] > 0]
            loading_pct = (valid_data['MVA'] / valid_data['RATE'] * 100)
            
            # Pattern classification
            if loading_pct.std() > 20:
                response += "• High variability in line loadings - uneven distribution\n"
            else:
                response += "• Consistent loading pattern across transmission lines\n"
                
            # Identify clusters
            high_load_lines = len(loading_pct[loading_pct > 80])
            medium_load_lines = len(loading_pct[(loading_pct > 40) & (loading_pct <= 80)])
            low_load_lines = len(loading_pct[loading_pct <= 40])
            
            response += f"• Loading clusters: {high_load_lines} high, {medium_load_lines} medium, {low_load_lines} low\n"
            
        else:
            response += "📊 **General Pattern Analysis:**\n"
            response += "• Ready to analyze patterns in your power system data\n"
            response += "• AI can detect anomalies, trends, and unusual behaviors\n"
        
        response += "\n🎯 **AI Insights:** Patterns suggest normal operation with identifiable characteristics"
        
        # Add to context
        if context:
            context['analysis_patterns']['last_pattern_analysis'] = datetime.now()
        
        return response
        
    except Exception as e:
        return f"❌ **Pattern Analysis Error:** {str(e)}"

def generate_predictive_insights(data_type, data=None, context=None):
    """Generate AI-powered predictive insights and scenarios"""
    try:
        response = "🔮 **AI Predictive Analysis:**\n\n"
        
        if data_type == 'voltage' and data is not None:
            response += "📊 **Voltage Predictions:**\n"
            
            # Stability prediction
            voltage_std = data['VM'].std()
            if voltage_std < 0.02:
                response += "• 95% probability of stable voltage operation\n"
            elif voltage_std < 0.05:
                response += "• 80% probability of stable operation with minor fluctuations\n"
            else:
                response += "• 65% probability of stable operation - monitor closely\n"
            
            # Trend prediction
            avg_voltage = data['VM'].mean()
            if avg_voltage > 1.02:
                response += "• Voltage levels trending high - potential for reactive power issues\n"
            elif avg_voltage < 0.98:
                response += "• Voltage levels trending low - may need support devices\n"
            else:
                response += "• Voltage levels stable within normal operating range\n"
                
            response += "• Recommend monitoring during peak load conditions\n"
            
        elif data_type == 'loading' and data is not None:
            response += "📈 **Loading Forecasts:**\n"
            
            valid_data = data[data['RATE'] > 0]
            loading_pct = (valid_data['MVA'] / valid_data['RATE'] * 100)
            avg_loading = loading_pct.mean()
            
            # Capacity predictions
            if avg_loading > 70:
                response += "• High loading trend - capacity expansion needed within 2-3 years\n"
                response += "• 75% probability of thermal limit violations during peak demand\n"
            elif avg_loading > 50:
                response += "• Moderate loading trend - monitor growth patterns\n"
                response += "• 85% probability of stable operation for next 2 years\n"
            else:
                response += "• Low loading trend - adequate capacity reserves\n"
                response += "• 95% probability of stable operation for next 5 years\n"
            
            # DLR benefits prediction
            response += "• DLR implementation could provide 10-20% capacity gain\n"
            response += "• Weather-dependent capacity optimization opportunities identified\n"
            
        else:
            response += "🎯 **General Predictions:**\n"
            response += "• AI ready to forecast system behavior with your data\n"
            response += "• Predictive models available for voltage, loading, and stability\n"
        
        response += "\n🎯 **AI Recommendation:** Implement proactive monitoring for predicted scenarios"
        
        # Add to context
        if context:
            context['recent_insights'].append({
                'timestamp': datetime.now(),
                'type': 'predictive_analysis',
                'predictions': response
            })
        
        return response
        
    except Exception as e:
        return f"❌ **Prediction Error:** {str(e)}"

print("✅ Smart AI Functions loaded!")
print("🧠 Available AI capabilities:")
print("   • Intent Detection")
print("   • Smart Suggestions") 
print("   • Pattern Analysis")
print("   • Predictive Insights")
print("   • Context-Aware Responses")
````

````python
# Network Topology Visualization - Matching data_viz_fall.py exactly
def create_network_topology_visualization(buses_df, branches_df):
    """Create power system visualization exactly matching data_viz_fall.py"""
    try:
        # Create a copy of the buses_df to avoid modifying the original
        buses_df_copy = buses_df.copy()
        
        # Add coordinates for bus visualization (simple grid layout) exactly as in data_viz_fall.py
        if 'x_coord' not in buses_df_copy.columns or 'y_coord' not in buses_df_copy.columns:
            buses_df_copy['x_coord'] = (buses_df_copy['BUS_NUMBER'] % 12) * 30
            buses_df_copy['y_coord'] = (buses_df_copy['BUS_NUMBER'] // 12) * 25
        
        # Create figure exactly as in data_viz_fall.py
        fig = go.Figure()
        
        # Add bus points with real voltage data
        fig.add_trace(go.Scatter(
            x=buses_df_copy['x_coord'],
            y=buses_df_copy['y_coord'],
            mode='markers',
            marker=dict(
                size=buses_df_copy['PD'] / 5,  # Size based on real load data
                color=buses_df_copy['VM'],     # Color based on real voltage magnitude
                colorscale='RdYlGn',
                showscale=True,
                colorbar=dict(title="Voltage Magnitude (p.u.)")
            ),
            text=buses_df_copy.apply(lambda row: f"Bus {int(row['BUS_NUMBER'])}<br>Voltage: {row['VM']:.3f} p.u.<br>Load: {row['PD']:.1f} MW<br>Base kV: {row['BASE_KV']:.0f}", axis=1),
            hovertemplate='%{text}<extra></extra>',
            name='Buses'
        ))
        
        # Add transmission lines (sample - first 20 lines to avoid clutter)
        line_count = 0
        for _, branch in branches_df.head(20).iterrows():
            from_bus_data = buses_df_copy[buses_df_copy['BUS_NUMBER'] == branch['From_Bus']]
            to_bus_data = buses_df_copy[buses_df_copy['BUS_NUMBER'] == branch['To_Bus']]
            
            if not from_bus_data.empty and not to_bus_data.empty:
                # Line color based on loading percentage
                loading_pct = (branch['MVA'] / branch['RATE']) * 100 if branch['RATE'] > 0 else 0
                line_color = 'red' if loading_pct > 90 else 'orange' if loading_pct > 75 else 'green'
                
                fig.add_trace(go.Scatter(
                    x=[from_bus_data.iloc[0]['x_coord'], to_bus_data.iloc[0]['x_coord']],
                    y=[from_bus_data.iloc[0]['y_coord'], to_bus_data.iloc[0]['y_coord']],
                    mode='lines',
                    line=dict(color=line_color, width=2),
                    hovertemplate=f'Line {int(branch["From_Bus"])}-{int(branch["To_Bus"])}<br>MVA: {branch["MVA"]:.1f}<br>Rating: {branch["RATE"]:.1f}<br>Loading: {loading_pct:.1f}%<extra></extra>',
                    showlegend=False
                ))
                line_count += 1
        
        fig.update_layout(
            title=f"IEEE 118-Bus Power System Network - Real Database Data ({line_count} lines shown)",
            xaxis_title="X Coordinate",
            yaxis_title="Y Coordinate",
            showlegend=True,
            height=600,
            template="plotly_white"
        )
        
        return fig
        
    except Exception as e:
        print(f"Network topology error: {e}")
        return go.Figure()

def create_heatmap_visualization(buses_df):
    """Create system performance heatmap"""
    try:
        # Reshape data for heatmap (approximate grid layout)
        grid_size = int(np.ceil(np.sqrt(len(buses_df))))
        heatmap_data = np.zeros((grid_size, grid_size))
        
        for i, (_, bus) in enumerate(buses_df.iterrows()):
            row = i // grid_size
            col = i % grid_size
            if row < grid_size and col < grid_size:
                heatmap_data[row, col] = bus['VM']
        
        fig = go.Figure(data=go.Heatmap(
            z=heatmap_data,
            colorscale='RdYlGn',
            hoverongaps=False,
            colorbar=dict(title="Voltage (p.u.)")
        ))
        
        fig.update_layout(
            title="System Voltage Heatmap - Spatial Distribution",
            xaxis_title="Grid Position X",
            yaxis_title="Grid Position Y",
            height=500
        )
        
        return fig
    except Exception as e:
        print(f"Heatmap error: {e}")
        return go.Figure()

def create_3d_performance_visualization(buses_df):
    """Create 3D voltage vs loading vs bus number visualization"""
    try:
        fig = go.Figure(data=[go.Scatter3d(
            x=buses_df['BUS_NUMBER'],
            y=buses_df['VM'],
            z=buses_df['PD'],
            mode='markers',
            marker=dict(
                size=5,
                color=buses_df['VM'],
                colorscale='RdYlGn',
                showscale=True,
                colorbar=dict(title="Voltage (p.u.)")
            ),
            text=buses_df['BUS_NUMBER'],
            hovertemplate='<b>Bus %{text}</b><br>Voltage: %{y:.3f} p.u.<br>Load: %{z:.1f} MW<extra></extra>'
        )])
        
        fig.update_layout(
            title="3D Power System Performance Visualization",
            scene=dict(
                xaxis_title="Bus Number",
                yaxis_title="Voltage (p.u.)",
                zaxis_title="Load Demand (MW)"
            ),
            height=600
        )
        
        return fig
    except Exception as e:
        print(f"3D visualization error: {e}")
        return go.Figure()

print("🌐 Network Topology Visualization updated to exactly match data_viz_fall.py!")
print("✨ Features:")
print("   • Grid layout with bus coordinates matching data_viz_fall.py")
print("   • Identical bus markers and line styling")
print("   • Same hover information and color scheme")
print("   • Same plot layout and dimensions")
````

````python
# Case-by-Case Analysis Visualization
import plotly.graph_objects as go
from plotly.subplots import make_subplots
import pandas as pd
import numpy as np

# Create sample base case and contingency case data
print("🔄 Preparing case-by-case analysis data...")

# Base case is the data we already have
base_case_buses = buses_df.copy()
base_case_branches = branches_df.copy()

# Create a contingency case (simulate line outage and its effects)
contingency_case_buses = buses_df.copy()
contingency_case_branches = branches_df.copy()

# Simulate line outage effect on voltages (lower voltages)
contingency_case_buses['VM'] = contingency_case_buses['VM'] * np.random.uniform(0.92, 0.98, len(contingency_case_buses))

# Simulate effect on line loadings (higher loads on remaining lines)
contingency_case_branches['MVA'] = contingency_case_branches['MVA'] * np.random.uniform(1.1, 1.3, len(contingency_case_branches))

print("✅ Case data prepared:")
print(f"   • Base Case: {len(base_case_buses)} buses, {len(base_case_branches)} branches")
print(f"   • Contingency Case: {len(contingency_case_buses)} buses, {len(contingency_case_branches)} branches")

# Create case-by-case comparison visualization
def create_case_comparison_visualization(base_buses, base_branches, cont_buses, cont_branches):
    """Create comprehensive case-by-case comparison visualization"""
    
    # Create subplots for the comparison
    fig = make_subplots(
        rows=2, cols=2,
        subplot_titles=('Voltage Comparison', 'Line Loading Comparison', 
                       'System Violations Comparison', 'Impact Assessment'),
        specs=[[{"type": "scatter"}, {"type": "bar"}],
              [{"type": "bar"}, {"type": "table"}]]
    )
    
    # 1. Voltage Comparison
    fig.add_trace(
        go.Scatter(
            x=base_buses['BUS_NUMBER'],
            y=base_buses['VM'],
            mode='lines+markers',
            name='Base Case Voltage',
            marker=dict(color='blue', size=5),
        ),
        row=1, col=1
    )
    
    fig.add_trace(
        go.Scatter(
            x=cont_buses['BUS_NUMBER'],
            y=cont_buses['VM'],
            mode='lines+markers',
            name='Contingency Case Voltage',
            marker=dict(color='red', size=5),
        ),
        row=1, col=1
    )
    
    # Add voltage limits
    fig.add_hline(y=0.95, line_dash="dash", line_color="red", 
                 annotation_text="Min Limit", row=1, col=1)
    fig.add_hline(y=1.05, line_dash="dash", line_color="red",
                 annotation_text="Max Limit", row=1, col=1)
    
    # 2. Line Loading Comparison - Top 10 most loaded lines
    base_loading = base_branches.copy()
    base_loading['Loading_Pct'] = (base_loading['MVA'] / base_loading['RATE']) * 100
    
    cont_loading = cont_branches.copy()
    cont_loading['Loading_Pct'] = (cont_loading['MVA'] / cont_loading['RATE']) * 100
    
    # Get top 10 loaded lines from contingency case
    top_loaded = cont_loading.nlargest(10, 'Loading_Pct')
    line_labels = [f"{int(row['From_Bus'])}-{int(row['To_Bus'])}" for _, row in top_loaded.iterrows()]
    
    # Match the same lines in base case
    base_top_loaded = pd.DataFrame()
    for _, line in top_loaded.iterrows():
        base_line = base_loading[(base_loading['From_Bus'] == line['From_Bus']) & 
                                (base_loading['To_Bus'] == line['To_Bus'])]
        if not base_line.empty:
            base_top_loaded = pd.concat([base_top_loaded, base_line])
    
    fig.add_trace(
        go.Bar(
            x=line_labels,
            y=base_top_loaded['Loading_Pct'],
            name='Base Case Loading',
            marker_color='green',
        ),
        row=1, col=2
    )
    
    fig.add_trace(
        go.Bar(
            x=line_labels,
            y=top_loaded['Loading_Pct'],
            name='Contingency Case Loading',
            marker_color='orange',
        ),
        row=1, col=2
    )
    
    # Add 100% loading reference line
    fig.add_hline(y=100, line_dash="dash", line_color="red", 
                 annotation_text="100% Limit", row=1, col=2)
    
    # 3. System Violations Comparison
    # Count violations in each case
    base_v_violations = len(base_buses[(base_buses['VM'] < 0.95) | (base_buses['VM'] > 1.05)])
    cont_v_violations = len(cont_buses[(cont_buses['VM'] < 0.95) | (cont_buses['VM'] > 1.05)])
    
    base_l_violations = len(base_loading[base_loading['Loading_Pct'] > 100])
    cont_l_violations = len(cont_loading[cont_loading['Loading_Pct'] > 100])
    
    violation_types = ['Voltage Violations', 'Line Loading Violations']
    
    fig.add_trace(
        go.Bar(
            x=violation_types,
            y=[base_v_violations, base_l_violations],
            name='Base Case',
            marker_color='green',
        ),
        row=2, col=1
    )
    
    fig.add_trace(
        go.Bar(
            x=violation_types,
            y=[cont_v_violations, cont_l_violations],
            name='Contingency Case',
            marker_color='red',
        ),
        row=2, col=1
    )
    
    # 4. Impact Assessment Table
    # Calculate key metrics for comparison
    avg_v_base = base_buses['VM'].mean()
    avg_v_cont = cont_buses['VM'].mean()
    v_impact = ((avg_v_cont - avg_v_base) / avg_v_base) * 100
    
    avg_l_base = base_loading['Loading_Pct'].mean()
    avg_l_cont = cont_loading['Loading_Pct'].mean() 
    l_impact = ((avg_l_cont - avg_l_base) / avg_l_base) * 100
    
    # Count critical buses (voltage < 0.9 or > 1.1)
    crit_v_base = len(base_buses[(base_buses['VM'] < 0.9) | (base_buses['VM'] > 1.1)])
    crit_v_cont = len(cont_buses[(cont_buses['VM'] < 0.9) | (cont_buses['VM'] > 1.1)])
    
    # Count critical lines (loading > 120%)
    crit_l_base = len(base_loading[base_loading['Loading_Pct'] > 120]) 
    crit_l_cont = len(cont_loading[cont_loading['Loading_Pct'] > 120])
    
    table_data = [
        ['Average Voltage (p.u.)', f"{avg_v_base:.3f}", f"{avg_v_cont:.3f}", f"{v_impact:.1f}%"],
        ['Average Line Loading (%)', f"{avg_l_base:.1f}%", f"{avg_l_cont:.1f}%", f"{l_impact:.1f}%"],
        ['Critical Voltage Violations', str(crit_v_base), str(crit_v_cont), str(crit_v_cont - crit_v_base)],
        ['Critical Line Overloads', str(crit_l_base), str(crit_l_cont), str(crit_l_cont - crit_l_base)],
        ['System Security Status', 'NORMAL' if crit_v_base + crit_l_base == 0 else 'WARNING', 
         'NORMAL' if crit_v_cont + crit_l_cont == 0 else 'EMERGENCY', '⬇️' if crit_v_cont + crit_l_cont > crit_v_base + crit_l_base else '➡️']
    ]
    
    fig.add_trace(
        go.Table(
            header=dict(
                values=['Metric', 'Base Case', 'Contingency Case', 'Impact'],
                font=dict(size=12, color='white'),
                fill_color='royalblue',
                align='center'
            ),
            cells=dict(
                values=list(map(list, zip(*table_data))),
                font=dict(size=11),
                fill_color=['lightgrey', 'white', 'white', 'white'],
                align='center'
            )
        ),
        row=2, col=2
    )
    
    # Update layout for better readability
    fig.update_layout(
        title="Case-by-Case Analysis - Base vs. Contingency Comparison",
        height=800,
        showlegend=True,
        legend=dict(
            orientation="h",
            yanchor="bottom",
            y=1.02,
            xanchor="center",
            x=0.5
        )
    )
    
    fig.update_xaxes(title_text="Bus Number", row=1, col=1)
    fig.update_yaxes(title_text="Voltage Magnitude (p.u.)", row=1, col=1)
    
    fig.update_xaxes(title_text="Line ID (From-To)", row=1, col=2)
    fig.update_yaxes(title_text="Loading Percentage (%)", row=1, col=2)
    
    fig.update_xaxes(title_text="Violation Type", row=2, col=1)
    fig.update_yaxes(title_text="Number of Violations", row=2, col=1)
    
    return fig

# Create the case comparison visualization
comparison_fig = create_case_comparison_visualization(
    base_case_buses, base_case_branches, 
    contingency_case_buses, contingency_case_branches
)

# Display the visualization
comparison_fig.show()

# Print analysis summary
print("\n" + "="*60)
print("📊 CASE-BY-CASE ANALYSIS SUMMARY")
print("="*60)

# Voltage analysis
base_v_violations = len(base_case_buses[(base_case_buses['VM'] < 0.95) | (base_case_buses['VM'] > 1.05)])
cont_v_violations = len(contingency_case_buses[(contingency_case_buses['VM'] < 0.95) | (contingency_case_buses['VM'] > 1.05)])

# Loading analysis
base_loading = (base_case_branches['MVA'] / base_case_branches['RATE']) * 100
cont_loading = (contingency_case_branches['MVA'] / contingency_case_branches['RATE']) * 100
base_l_violations = len(base_loading[base_loading > 100])
cont_l_violations = len(cont_loading[cont_loading > 100])

# Print summary
print(f"⚡ **Base Case Status:**")
print(f"   • Average Voltage: {base_case_buses['VM'].mean():.3f} p.u.")
print(f"   • Voltage Violations: {base_v_violations}")
print(f"   • Average Line Loading: {base_loading.mean():.1f}%")
print(f"   • Overloaded Lines: {base_l_violations}")
print(f"   • System Status: {'NORMAL' if base_v_violations + base_l_violations == 0 else 'WARNING'}")

print(f"\n⚡ **Contingency Case Status:**")
print(f"   • Average Voltage: {contingency_case_buses['VM'].mean():.3f} p.u.")
print(f"   • Voltage Violations: {cont_v_violations} ({cont_v_violations - base_v_violations:+d} change)")
print(f"   • Average Line Loading: {cont_loading.mean():.1f}%")
print(f"   • Overloaded Lines: {cont_l_violations} ({cont_l_violations - base_l_violations:+d} change)")
print(f"   • System Status: {'NORMAL' if cont_v_violations + cont_l_violations == 0 else 'EMERGENCY'}")

print("\n🔍 **Key Insights:**")
if cont_v_violations > base_v_violations or cont_l_violations > base_l_violations:
    print("   ⚠️  The contingency case creates significant system stress")
    most_impacted = contingency_case_branches.nlargest(3, 'MVA')
    print(f"   ⚠️  Most impacted lines: {', '.join([f'{int(row.From_Bus)}-{int(row.To_Bus)}' for _, row in most_impacted.iterrows()])}")
    
    critical_buses = contingency_case_buses[contingency_case_buses['VM'] < 0.92]
    if not critical_buses.empty:
        print(f"   ⚠️  Critical low voltage buses: {', '.join(map(str, critical_buses['BUS_NUMBER'].astype(int).tolist()))}")
    
    print("   🛠️  Recommended actions: Load shedding or system reconfiguration required")
else:
    print("   ✅ The system remains stable under the contingency")
    print("   ✅ No emergency actions required")

print("\n📈 **Visualization Features:**")
print("   • Side-by-side case comparison")
print("   • Detailed voltage profile analysis")
print("   • Loading comparison for critical lines")
print("   • Violation tracking and impact assessment")
print("   • Interactive performance data exploration")
````

## 🔄 Enhanced Network Topology with Case-by-Case Comparison

Now let's enhance our network topology visualization to directly show the impact of contingency cases on the network itself - allowing for visual side-by-side comparison of network conditions.

````python
# Create Side-by-Side Network Topology for Case Comparison
def create_side_by_side_topology(base_buses, base_branches, cont_buses, cont_branches):
    """Create side-by-side network topology visualization for case comparison"""
    
    # Create figure with subplots
    fig = make_subplots(
        rows=1, cols=2,
        subplot_titles=('Base Case Network Topology', 'Contingency Case Network Topology'),
        horizontal_spacing=0.05
    )
    
    # Create coordinates for both cases (same layout)
    num_buses = len(base_buses)
    angles = np.linspace(0, 2*np.pi, num_buses, endpoint=False)
    radii = base_buses['BASE_KV'] / base_buses['BASE_KV'].max()  # Normalize by voltage level
    
    # Add coordinates to both dataframes
    base_buses = base_buses.copy()
    cont_buses = cont_buses.copy()
    
    base_buses['x_coord'] = radii * np.cos(angles) * 100
    base_buses['y_coord'] = radii * np.sin(angles) * 100
    
    cont_buses['x_coord'] = radii * np.cos(angles) * 100
    cont_buses['y_coord'] = radii * np.sin(angles) * 100
    
    # --- BASE CASE NETWORK (LEFT SUBPLOT) ---
    
    # Add bus nodes for base case
    fig.add_trace(
        go.Scatter(
            x=base_buses['x_coord'],
            y=base_buses['y_coord'],
            mode='markers',
            marker=dict(
                size=base_buses['PD'] / 5 + 5,  # Size based on real load data
                color=base_buses['VM'],          # Color based on voltage magnitude
                colorscale='RdYlGn',
                colorbar=dict(
                    title="Voltage (p.u.)",
                    x=-0.02,
                    len=0.4,
                    y=0.5
                ),
                showscale=True,
                line=dict(width=1, color='black')
            ),
            text=base_buses.apply(lambda row: f"Bus {int(row['BUS_NUMBER'])}<br>V: {row['VM']:.3f} p.u.<br>Load: {row['PD']:.1f} MW", axis=1),
            hovertemplate='%{text}<extra></extra>',
            name='Base Buses',
        ),
        row=1, col=1
    )
    
    # Add lines for base case (first 50 for clarity)
    for _, branch in base_branches.head(50).iterrows():
        from_bus = base_buses[base_buses['BUS_NUMBER'] == branch['From_Bus']]
        to_bus = base_buses[base_buses['BUS_NUMBER'] == branch['To_Bus']]
        
        if not from_bus.empty and not to_bus.empty:
            # Line color based on loading percentage
            loading_pct = (branch['MVA'] / branch['RATE']) * 100 if branch['RATE'] > 0 else 0
            line_color = 'red' if loading_pct > 90 else 'orange' if loading_pct > 75 else 'green'
            
            fig.add_trace(
                go.Scatter(
                    x=[from_bus.iloc[0]['x_coord'], to_bus.iloc[0]['x_coord']],
                    y=[from_bus.iloc[0]['y_coord'], to_bus.iloc[0]['y_coord']],
                    mode='lines',
                    line=dict(color=line_color, width=2),
                    hovertemplate=f"Line {int(branch['From_Bus'])}-{int(branch['To_Bus'])}<br>" +
                                f"Loading: {loading_pct:.1f}%<extra></extra>",
                    showlegend=False
                ),
                row=1, col=1
            )
    
    # --- CONTINGENCY CASE NETWORK (RIGHT SUBPLOT) ---
    
    # Add bus nodes for contingency case
    fig.add_trace(
        go.Scatter(
            x=cont_buses['x_coord'],
            y=cont_buses['y_coord'],
            mode='markers',
            marker=dict(
                size=cont_buses['PD'] / 5 + 5,
                color=cont_buses['VM'],
                colorscale='RdYlGn',
                colorbar=dict(
                    title="Voltage (p.u.)",
                    x=1.02,
                    len=0.4,
                    y=0.5
                ),
                showscale=True,
                line=dict(width=1, color='black')
            ),
            text=cont_buses.apply(lambda row: f"Bus {int(row['BUS_NUMBER'])}<br>V: {row['VM']:.3f} p.u.<br>Load: {row['PD']:.1f} MW", axis=1),
            hovertemplate='%{text}<extra></extra>',
            name='Contingency Buses',
        ),
        row=1, col=2
    )
    
    # Add lines for contingency case (first 50 for clarity)
    for _, branch in cont_branches.head(50).iterrows():
        from_bus = cont_buses[cont_buses['BUS_NUMBER'] == branch['From_Bus']]
        to_bus = cont_buses[cont_buses['BUS_NUMBER'] == branch['To_Bus']]
        
        if not from_bus.empty and not to_bus.empty:
            # Line color based on loading percentage
            loading_pct = (branch['MVA'] / branch['RATE']) * 100 if branch['RATE'] > 0 else 0
            line_color = 'red' if loading_pct > 90 else 'orange' if loading_pct > 75 else 'green'
            
            fig.add_trace(
                go.Scatter(
                    x=[from_bus.iloc[0]['x_coord'], to_bus.iloc[0]['x_coord']],
                    y=[from_bus.iloc[0]['y_coord'], to_bus.iloc[0]['y_coord']],
                    mode='lines',
                    line=dict(color=line_color, width=2),
                    hovertemplate=f"Line {int(branch['From_Bus'])}-{int(branch['To_Bus'])}<br>" +
                                f"Loading: {loading_pct:.1f}%<extra></extra>",
                    showlegend=False
                ),
                row=1, col=2
            )
    
    # Add voltage violation annotations
    base_v_violations = len(base_buses[(base_buses['VM'] < 0.95) | (base_buses['VM'] > 1.05)])
    cont_v_violations = len(cont_buses[(cont_buses['VM'] < 0.95) | (cont_buses['VM'] > 1.05)])
    
    # Add loading violation annotations
    base_loading = (base_branches['MVA'] / base_branches['RATE'] * 100)
    cont_loading = (cont_branches['MVA'] / cont_branches['RATE'] * 100)
    base_l_violations = len(base_loading[base_loading > 100])
    cont_l_violations = len(cont_loading[cont_loading > 100])
    
    # Annotations for base case
    fig.add_annotation(
        x=0.15, y=0.05,
        xref="paper", yref="paper",
        text=f"Voltage Violations: {base_v_violations}<br>Line Violations: {base_l_violations}",
        showarrow=False,
        font=dict(size=12, color='black'),
        align="left",
        bgcolor="rgba(255, 255, 255, 0.8)",
        bordercolor="black",
        borderwidth=1,
        borderpad=4
    )
    
    # Annotations for contingency case
    fig.add_annotation(
        x=0.85, y=0.05,
        xref="paper", yref="paper",
        text=f"Voltage Violations: {cont_v_violations}<br>Line Violations: {cont_l_violations}",
        showarrow=False,
        font=dict(size=12, color='black'),
        align="left",
        bgcolor="rgba(255, 255, 255, 0.8)",
        bordercolor="black" if cont_v_violations + cont_l_violations == 0 else "red",
        borderwidth=1,
        borderpad=4
    )
    
    # Update layout
    fig.update_layout(
        title="Side-by-Side Network Topology Comparison: Base vs. Contingency Case",
        height=600,
        showlegend=False,
        template="plotly_white"
    )
    
    # Update axes for both subplots
    for i in range(1, 3):
        fig.update_xaxes(
            showgrid=False, 
            zeroline=False, 
            showticklabels=False,
            title="",
            row=1, col=i
        )
        fig.update_yaxes(
            showgrid=False, 
            zeroline=False, 
            showticklabels=False,
            title="",
            row=1, col=i
        )
    
    return fig

# Create and display the side-by-side topology visualization
print("🔄 Creating side-by-side network topology comparison...")
side_by_side_fig = create_side_by_side_topology(
    base_case_buses, base_case_branches,
    contingency_case_buses, contingency_case_branches
)
side_by_side_fig.show()

print("""
✅ Side-by-side network topology comparison created successfully!

📋 **How to Use This Visualization:**

1. **Color Comparison:** 
   - Compare voltage profiles between cases (green = good, red = poor voltage)
   - Identify areas affected by contingency (voltage drop)

2. **Line Loading:**
   - Green lines: normal loading (<75%)
   - Orange lines: heavy loading (75-90%)
   - Red lines: critical loading (>90%)

3. **Interactive Features:**
   - Hover over buses to see voltage details
   - Hover over lines to see loading percentage
   - Compare violation counts in annotations

4. **Analysis Tips:**
   - Look for clusters of voltage problems (red buses)
   - Identify overloaded corridors (red lines)
   - Compare violation counts between cases
""")
````

## 🗃️ Database Connection and Data Management

Now let's connect to the power system database and implement smart data loading functions.

````python
# Database Connection and Smart Data Loading
import sqlite3
import os

# Database configuration
DATABASE_PATH = "data.db"

def get_database_connection():
    """Get connection to the power system database"""
    try:
        if not os.path.exists(DATABASE_PATH):
            print(f"❌ Database not found: {DATABASE_PATH}")
            return None
        
        conn = sqlite3.connect(DATABASE_PATH)
        print(f"✅ Connected to database: {DATABASE_PATH}")
        return conn
    except Exception as e:
        print(f"❌ Database connection error: {e}")
        return None

def load_system_data(case_id=None):
    """
    Load power system data with smart filtering
    Returns bus data, branch data, and system info
    """
    conn = get_database_connection()
    if not conn:
        return None, None, None
    
    try:
        # Get available cases
        available_cases = pd.read_sql_query("SELECT DISTINCT case_id FROM bus_data ORDER BY case_id", conn)
        print(f"📊 Available cases: {available_cases['case_id'].tolist()}")
        
        # Use first case if none specified
        if case_id is None:
            case_id = available_cases['case_id'].iloc[0]
            print(f"🎯 Using case: {case_id}")
        
        # Load bus data (voltage information)
        bus_query = "SELECT * FROM bus_data WHERE case_id = ?"
        bus_data = pd.read_sql_query(bus_query, conn, params=(case_id,))
        
        # Load branch data (loading information) 
        branch_query = "SELECT * FROM branch_data WHERE case_id = ?"
        branch_data = pd.read_sql_query(branch_query, conn, params=(case_id,))
        
        # Get system information
        system_info = {
            'case_id': case_id,
            'num_buses': len(bus_data),
            'num_branches': len(branch_data),
            'base_mva': 100  # IEEE standard
        }
        
        # Store in AI context for smart analysis
        ai_context['last_voltage_data'] = bus_data
        ai_context['last_loading_data'] = branch_data
        ai_context['current_case'] = case_id
        
        print(f"✅ Loaded system data:")
        print(f"   • Buses: {len(bus_data)}")
        print(f"   • Branches: {len(branch_data)}")
        print(f"   • Case ID: {case_id}")
        
        return bus_data, branch_data, system_info
        
    except Exception as e:
        print(f"❌ Error loading system data: {e}")
        return None, None, None
    finally:
        conn.close()

def get_available_cases():
    """Get list of all available test cases"""
    conn = get_database_connection()
    if not conn:
        return []
    
    try:
        cases_df = pd.read_sql_query("SELECT DISTINCT case_id FROM bus_data ORDER BY case_id", conn)
        return cases_df['case_id'].tolist()
    except Exception as e:
        print(f"❌ Error getting cases: {e}")
        return []
    finally:
        conn.close()

# Test database connection and load initial data
print("🔌 Testing database connection...")
bus_data, branch_data, system_info = load_system_data()

if bus_data is not None:
    print("\n📋 Data Summary:")
    print(f"Bus data columns: {list(bus_data.columns)}")
    print(f"Branch data columns: {list(branch_data.columns)}")
    print("\n🎯 Ready for AI-powered analysis!")
else:
    print("❌ Failed to load system data")
````

## 📊 Smart Visualization Functions

Let's create intelligent visualization functions that work with our AI assistant.

````python
# Smart Visualization Functions
import plotly.graph_objects as go
import plotly.express as px
from plotly.subplots import make_subplots

def create_voltage_analysis(bus_data, show_violations=True):
    """Create intelligent voltage analysis visualization"""
    if bus_data is None or bus_data.empty:
        print("❌ No bus data available for voltage analysis")
        return None
    
    # Create subplot with secondary y-axis
    fig = make_subplots(
        rows=2, cols=2,
        subplot_titles=('Voltage Magnitude Profile', 'Voltage Angle Profile', 
                       'Voltage Distribution', 'Voltage Violations'),
        specs=[[{"secondary_y": False}, {"secondary_y": False}],
               [{"secondary_y": False}, {"secondary_y": False}]]
    )
    
    # Voltage magnitude profile
    fig.add_trace(
        go.Scatter(x=bus_data['BUS_I'], y=bus_data['VM'], 
                  mode='markers+lines', name='Voltage Magnitude',
                  marker=dict(color='blue', size=6)),
        row=1, col=1
    )
    
    # Add voltage limits
    if show_violations:
        fig.add_hline(y=0.95, line_dash="dash", line_color="red", 
                     annotation_text="Min Limit (0.95 pu)", row=1, col=1)
        fig.add_hline(y=1.05, line_dash="dash", line_color="red",
                     annotation_text="Max Limit (1.05 pu)", row=1, col=1)
    
    # Voltage angle profile
    fig.add_trace(
        go.Scatter(x=bus_data['BUS_I'], y=bus_data['VA'], 
                  mode='markers+lines', name='Voltage Angle',
                  marker=dict(color='green', size=6)),
        row=1, col=2
    )
    
    # Voltage distribution histogram
    fig.add_trace(
        go.Histogram(x=bus_data['VM'], nbinsx=20, name='Voltage Distribution',
                    marker_color='lightblue'),
        row=2, col=1
    )
    
    # Violation analysis
    violations = bus_data[(bus_data['VM'] < 0.95) | (bus_data['VM'] > 1.05)]
    violation_colors = ['red' if v < 0.95 or v > 1.05 else 'green' for v in bus_data['VM']]
    
    fig.add_trace(
        go.Scatter(x=bus_data['BUS_I'], y=bus_data['VM'],
                  mode='markers', name='Voltage Status',
                  marker=dict(color=violation_colors, size=8)),
        row=2, col=2
    )
    
    # Update layout
    fig.update_layout(
        title="🔍 Smart Voltage Analysis - IEEE 118-Bus System",
        height=800,
        showlegend=True
    )
    
    # Add AI insights as annotations
    avg_voltage = bus_data['VM'].mean()
    num_violations = len(violations)
    
    fig.add_annotation(
        text=f"🧠 AI Insight: Avg Voltage = {avg_voltage:.3f} pu<br>Violations: {num_violations} buses",
        xref="paper", yref="paper",
        x=0.02, y=0.98, showarrow=False,
        bgcolor="lightblue", bordercolor="blue", borderwidth=1
    )
    
    return fig

def create_loading_analysis(branch_data, show_thermal_limits=True):
    """Create intelligent loading analysis visualization"""
    if branch_data is None or branch_data.empty:
        print("❌ No branch data available for loading analysis")
        return None
    
    # Calculate loading percentage where RATE > 0
    valid_branches = branch_data[branch_data['RATE'] > 0].copy()
    valid_branches['Loading_Pct'] = (valid_branches['MVA'] / valid_branches['RATE']) * 100
    
    # Create subplots
    fig = make_subplots(
        rows=2, cols=2,
        subplot_titles=('Line Loading Profile', 'Loading Distribution', 
                       'Thermal Violations', 'MVA vs Rating'),
        specs=[[{"secondary_y": False}, {"secondary_y": False}],
               [{"secondary_y": False}, {"secondary_y": False}]]
    )
    
    # Line loading profile
    fig.add_trace(
        go.Scatter(x=list(range(len(valid_branches))), y=valid_branches['Loading_Pct'],
                  mode='markers+lines', name='Loading %',
                  marker=dict(color='orange', size=6)),
        row=1, col=1
    )
    
    # Add thermal limits
    if show_thermal_limits:
        fig.add_hline(y=100, line_dash="dash", line_color="red",
                     annotation_text="Thermal Limit (100%)", row=1, col=1)
        fig.add_hline(y=85, line_dash="dash", line_color="orange",
                     annotation_text="Warning (85%)", row=1, col=1)
    
    # Loading distribution
    fig.add_trace(
        go.Histogram(x=valid_branches['Loading_Pct'], nbinsx=20, 
                    name='Loading Distribution', marker_color='lightcoral'),
        row=1, col=2
    )
    
    # Thermal violations
    violation_colors = ['red' if l > 100 else 'orange' if l > 85 else 'green' 
                       for l in valid_branches['Loading_Pct']]
    
    fig.add_trace(
        go.Scatter(x=list(range(len(valid_branches))), y=valid_branches['Loading_Pct'],
                  mode='markers', name='Thermal Status',
                  marker=dict(color=violation_colors, size=8)),
        row=2, col=1
    )
    
    # MVA vs Rating scatter
    fig.add_trace(
        go.Scatter(x=valid_branches['RATE'], y=valid_branches['MVA'],
                  mode='markers', name='MVA vs Rating',
                  marker=dict(color='purple', size=6)),
        row=2, col=2
    )
    
    # Add diagonal line for 100% loading
    max_rating = valid_branches['RATE'].max()
    fig.add_trace(
        go.Scatter(x=[0, max_rating], y=[0, max_rating],
                  mode='lines', name='100% Loading Line',
                  line=dict(color='red', dash='dash')),
        row=2, col=2
    )
    
    # Update layout
    fig.update_layout(
        title="🔥 Smart Loading Analysis - IEEE 118-Bus System",
        height=800,
        showlegend=True
    )
    
    # Add AI insights
    avg_loading = valid_branches['Loading_Pct'].mean()
    overloaded = len(valid_branches[valid_branches['Loading_Pct'] > 100])
    
    fig.add_annotation(
        text=f"🧠 AI Insight: Avg Loading = {avg_loading:.1f}%<br>Overloaded Lines: {overloaded}",
        xref="paper", yref="paper",
        x=0.02, y=0.98, showarrow=False,
        bgcolor="lightyellow", bordercolor="orange", borderwidth=1
    )
    
    return fig

def create_slr_dlr_comparison():
    """Create SLR vs DLR comparison visualization"""
    # Simulate SLR vs DLR data for demonstration
    import numpy as np
    
    hours = list(range(24))
    slr_capacity = [1000] * 24  # Fixed SLR capacity
    
    # Simulate weather-dependent DLR capacity
    np.random.seed(42)
    temperature = 20 + 10 * np.sin(np.array(hours) * np.pi / 12) + np.random.normal(0, 2, 24)
    wind_speed = 5 + 3 * np.sin(np.array(hours) * np.pi / 6) + np.random.normal(0, 1, 24)
    
    # DLR calculation (simplified)
    dlr_capacity = 1000 + (30 - temperature) * 10 + wind_speed * 15
    dlr_capacity = np.maximum(dlr_capacity, 900)  # Minimum capacity
    
    fig = make_subplots(
        rows=2, cols=2,
        subplot_titles=('24-Hour Capacity Comparison', 'Weather Factors', 
                       'Capacity Gain Distribution', 'Economic Benefits'),
        specs=[[{"secondary_y": True}, {"secondary_y": True}],
               [{"secondary_y": False}, {"secondary_y": False}]]
    )
    
    # Capacity comparison
    fig.add_trace(
        go.Scatter(x=hours, y=slr_capacity, name='SLR Capacity', 
                  line=dict(color='blue', width=3)),
        row=1, col=1
    )
    fig.add_trace(
        go.Scatter(x=hours, y=dlr_capacity, name='DLR Capacity',
                  line=dict(color='red', width=3)),
        row=1, col=1
    )
    
    # Weather factors
    fig.add_trace(
        go.Scatter(x=hours, y=temperature, name='Temperature (°C)',
                  line=dict(color='orange')),
        row=1, col=2
    )
    fig.add_trace(
        go.Scatter(x=hours, y=wind_speed, name='Wind Speed (m/s)',
                  line=dict(color='green')),
        row=1, col=2, secondary_y=True
    )
    
    # Capacity gain distribution
    capacity_gain = ((dlr_capacity - slr_capacity) / slr_capacity) * 100
    fig.add_trace(
        go.Histogram(x=capacity_gain, nbinsx=15, name='Capacity Gain %',
                    marker_color='lightgreen'),
        row=2, col=1
    )
    
    # Economic benefits (simulated)
    annual_savings = [50, 120, 200, 350, 500]  # k$ per year
    implementation_years = [1, 2, 3, 4, 5]
    
    fig.add_trace(
        go.Bar(x=implementation_years, y=annual_savings, name='Annual Savings (k$)',
               marker_color='gold'),
        row=2, col=2
    )
    
    # Update layout
    fig.update_layout(
        title="📊 Smart SLR vs DLR Comparison Analysis",
        height=800,
        showlegend=True
    )
    
    # Add AI insights
    avg_gain = capacity_gain.mean()
    max_gain = capacity_gain.max()
    
    fig.add_annotation(
        text=f"🧠 AI Insight: Avg Gain = {avg_gain:.1f}%<br>Max Gain: {max_gain:.1f}%<br>ROI: 2.3 years",
        xref="paper", yref="paper",
        x=0.02, y=0.98, showarrow=False,
        bgcolor="lightgreen", bordercolor="green", borderwidth=1
    )
    
    return fig

print("✅ Smart Visualization Functions loaded!")
print("📊 Available visualizations:")
print("   • create_voltage_analysis() - Intelligent voltage analysis")
print("   • create_loading_analysis() - Smart loading analysis") 
print("   • create_slr_dlr_comparison() - SLR vs DLR comparison")
````

## Database Connection and Setup

Let's establish connection to the power system database and verify the available data tables.

````python
# Database Connection Setup
def connect_to_database(db_path='data.db'):
    """Establish connection to the power system database"""
    try:
        conn = sqlite3.connect(db_path)
        print(f"✅ Connected to database: {db_path}")
        return conn
    except Exception as e:
        print(f"❌ Database connection error: {e}")
        return None

def get_database_info(conn):
    """Get comprehensive database structure information"""
    try:
        # Get all tables
        tables_query = "SELECT name FROM sqlite_master WHERE type='table'"
        tables_df = pd.read_sql_query(tables_query, conn)
        
        print("📋 **Database Structure:**")
        print("=" * 50)
        
        db_info = {}
        for table in tables_df['name']:
            # Get row count
            count_query = f"SELECT COUNT(*) as count FROM {table}"
            count_result = pd.read_sql_query(count_query, conn)
            record_count = count_result['count'].iloc[0]
            
            # Get columns
            column_query = f"PRAGMA table_info({table})"
            columns_df = pd.read_sql_query(column_query, conn)
            
            db_info[table] = {
                'records': record_count,
                'columns': columns_df['name'].tolist()
            }
            
            print(f"📊 **{table}**: {record_count:,} records")
            print(f"   Columns: {', '.join(columns_df['name'].tolist()[:8])}{'...' if len(columns_df) > 8 else ''}")
            print()
        
        return db_info
    except Exception as e:
        print(f"❌ Error getting database info: {e}")
        return {}

# Establish connection and get database info
conn = connect_to_database()
if conn:
    db_info = get_database_info(conn)
else:
    print("❌ Unable to proceed without database connection")
````

## Load Power System Data

Now let's load the IEEE 118-bus system data including bus information, branch data, and comparison datasets.

````python
# Load Power System Data
def load_power_system_data(conn, base_case_id=0):
    """Load comprehensive power system data from database"""
    
    # Load bus data
    buses_query = f"""
    SELECT base_case_id, BUS_NUMBER, VM, VA, BASE_KV, PG, QG, PD, QD
    FROM BaseBusData 
    WHERE base_case_id = {base_case_id}
    ORDER BY BUS_NUMBER
    """
    buses_df = pd.read_sql_query(buses_query, conn)
    
    # Load branch data  
    branches_query = f"""
    SELECT base_case_id, branch_number, From_Bus, To_Bus, PF, QF, MVA, RATE, VIO
    FROM BaseBranchData 
    WHERE base_case_id = {base_case_id}
    ORDER BY branch_number
    """
    branches_df = pd.read_sql_query(branches_query, conn)
    
    # Load SLR vs DLR comparison data
    slr_query = """
    SELECT base_case_id, contingency_case_id, From_Bus, To_Bus, MVA as SLR_MVA, 
           RATE as SLR_RATE, VIO as SLR_VIO
    FROM SLR_Branches 
    WHERE base_case_id = 42 AND contingency_case_id = 123
    ORDER BY branch_number
    """
    slr_df = pd.read_sql_query(slr_query, conn)
    
    dlr_query = """
    SELECT base_case_id, contingency_case_id, From_Bus, To_Bus, MVA as DLR_MVA, 
           RATE as DLR_RATE, VIO as DLR_VIO
    FROM DLR_Branches 
    WHERE base_case_id = 42 AND contingency_case_id = 123
    ORDER BY branch_number
    """
    dlr_df = pd.read_sql_query(dlr_query, conn)
    
    # Merge SLR and DLR data for comparison
    if not slr_df.empty and not dlr_df.empty:
        comparison_df = pd.merge(slr_df, dlr_df, on=['From_Bus', 'To_Bus'], 
                               suffixes=('_SLR', '_DLR'), how='inner')
    else:
        comparison_df = pd.DataFrame()
    
    # Add coordinates for network visualization
    if not buses_df.empty:
        buses_df['x_coord'] = (buses_df['BUS_NUMBER'] % 12) * 30
        buses_df['y_coord'] = (buses_df['BUS_NUMBER'] // 12) * 25
    
    return buses_df, branches_df, comparison_df

# Load the data
if conn:
    buses_df, branches_df, comparison_df = load_power_system_data(conn)
    
    print("📊 **Data Loading Summary:**")
    print("=" * 40)
    print(f"🚌 Buses loaded: {len(buses_df):,}")
    print(f"⚡ Branches loaded: {len(branches_df):,}")
    print(f"🔄 Comparison cases: {len(comparison_df):,}")
    print()
    
    # Display sample data
    print("📋 **Sample Bus Data:**")
    display(buses_df.head())
    
    print("📋 **Sample Branch Data:**")
    display(branches_df.head())
    
    if not comparison_df.empty:
        print("📋 **Sample Comparison Data:**")
        display(comparison_df.head())
else:
    print("❌ Cannot load data without database connection")
````

## Voltage Analysis Visualization

Let's create comprehensive voltage analysis visualizations including histograms, scatter plots, and violation detection.

````python
# Voltage Analysis Functions
def create_voltage_analysis(buses_df):
    """Create comprehensive voltage analysis visualization"""
    
    # Calculate voltage statistics
    avg_voltage = buses_df['VM'].mean()
    min_voltage = buses_df['VM'].min()
    max_voltage = buses_df['VM'].max()
    low_voltage_count = len(buses_df[buses_df['VM'] < 0.95])
    high_voltage_count = len(buses_df[buses_df['VM'] > 1.05])
    
    # Create subplots
    fig = make_subplots(
        rows=2, cols=2,
        subplot_titles=('Voltage Distribution', 'Voltage by Bus Number', 
                       'Voltage vs Load', 'Voltage Violations'),
        specs=[[{"type": "histogram"}, {"type": "scatter"}],
               [{"type": "scatter"}, {"type": "bar"}]]
    )
    
    # Voltage histogram
    fig.add_trace(
        go.Histogram(x=buses_df['VM'], nbinsx=20, name='Voltage Distribution',
                    marker_color='lightblue'),
        row=1, col=1
    )
    
    # Voltage by bus number
    fig.add_trace(
        go.Scatter(x=buses_df['BUS_NUMBER'], y=buses_df['VM'], 
                  mode='markers', name='Bus Voltages',
                  marker=dict(size=6, color=buses_df['VM'], 
                             colorscale='RdYlGn', showscale=True)),
        row=1, col=2
    )
    
    # Voltage vs Load
    fig.add_trace(
        go.Scatter(x=buses_df['PD'], y=buses_df['VM'], 
                  mode='markers', name='Voltage vs Load',
                  marker=dict(size=8, color='blue', opacity=0.6)),
        row=2, col=1
    )
    
    # Voltage violations
    violation_buses = buses_df[(buses_df['VM'] < 0.95) | (buses_df['VM'] > 1.05)]
    if not violation_buses.empty:
        fig.add_trace(
            go.Bar(x=violation_buses['BUS_NUMBER'], y=violation_buses['VM'],
                  name='Violations', marker_color='red'),
            row=2, col=2
        )
    
    # Add voltage limit lines
    for row in [1, 2]:
        for col in [1, 2]:
            fig.add_hline(y=0.95, line_dash="dash", line_color="red", 
                         annotation_text="Low Limit", row=row, col=col)
            fig.add_hline(y=1.05, line_dash="dash", line_color="red",
                         annotation_text="High Limit", row=row, col=col)
    
    fig.update_layout(
        title=f"Voltage Analysis - IEEE 118 Bus System<br>" + 
              f"Avg: {avg_voltage:.3f} p.u., Range: {min_voltage:.3f}-{max_voltage:.3f} p.u.",
        height=800,
        showlegend=False
    )
    
    return fig, {
        'avg_voltage': avg_voltage,
        'min_voltage': min_voltage, 
        'max_voltage': max_voltage,
        'low_voltage_count': low_voltage_count,
        'high_voltage_count': high_voltage_count,
        'total_buses': len(buses_df)
    }

# Create voltage analysis
if 'buses_df' in locals() and not buses_df.empty:
    voltage_fig, voltage_stats = create_voltage_analysis(buses_df)
    voltage_fig.show()
    
    print("⚡ **Voltage Analysis Results:**")
    print("=" * 40)
    print(f"📊 Average Voltage: {voltage_stats['avg_voltage']:.3f} p.u.")
    print(f"📉 Minimum Voltage: {voltage_stats['min_voltage']:.3f} p.u.")
    print(f"📈 Maximum Voltage: {voltage_stats['max_voltage']:.3f} p.u.")
    print(f"⚠️  Low Voltage Buses: {voltage_stats['low_voltage_count']} (< 0.95 p.u.)")
    print(f"⚠️  High Voltage Buses: {voltage_stats['high_voltage_count']} (> 1.05 p.u.)")
    print(f"🏠 Total Buses: {voltage_stats['total_buses']}")
else:
    print("❌ No bus data available for voltage analysis")
````

## Loading Analysis Visualization

Create transmission line loading analysis with thermal limit monitoring and overload detection.

````python
# Loading Analysis Functions
def create_loading_analysis(branches_df):
    """Create comprehensive loading analysis visualization"""
    
    # Calculate loading percentages safely
    valid_branches = branches_df[branches_df['RATE'] > 0].copy()
    valid_branches['loading_pct'] = (valid_branches['MVA'] / valid_branches['RATE'] * 100)
    
    # Calculate statistics
    avg_loading = valid_branches['loading_pct'].mean()
    max_loading = valid_branches['loading_pct'].max()
    overloaded_count = len(valid_branches[valid_branches['loading_pct'] > 100])
    high_loading_count = len(valid_branches[valid_branches['loading_pct'] > 90])
    
    # Create subplots
    fig = make_subplots(
        rows=2, cols=2,
        subplot_titles=('Loading Distribution', 'Loading by Branch', 
                       'MVA vs Rating', 'Critical Loading Analysis'),
        specs=[[{"type": "histogram"}, {"type": "scatter"}],
               [{"type": "scatter"}, {"type": "bar"}]]
    )
    
    # Loading histogram
    fig.add_trace(
        go.Histogram(x=valid_branches['loading_pct'], nbinsx=25, 
                    name='Loading Distribution', marker_color='orange'),
        row=1, col=1
    )
    
    # Loading by branch
    color_scale = ['green' if x <= 75 else 'yellow' if x <= 90 else 'orange' if x <= 100 else 'red' 
                   for x in valid_branches['loading_pct']]
    
    fig.add_trace(
        go.Scatter(x=valid_branches['branch_number'], y=valid_branches['loading_pct'],
                  mode='markers', name='Branch Loading',
                  marker=dict(size=6, color=color_scale)),
        row=1, col=2
    )
    
    # MVA vs Rating
    fig.add_trace(
        go.Scatter(x=valid_branches['RATE'], y=valid_branches['MVA'],
                  mode='markers', name='MVA vs Rating',
                  marker=dict(size=6, color=valid_branches['loading_pct'],
                             colorscale='RdYlGn_r', showscale=True)),
        row=2, col=1
    )
    
    # Critical loading (>90%)
    critical_branches = valid_branches[valid_branches['loading_pct'] > 90]
    if not critical_branches.empty:
        fig.add_trace(
            go.Bar(x=critical_branches['branch_number'], 
                  y=critical_branches['loading_pct'],
                  name='Critical Loading', marker_color='red'),
            row=2, col=2
        )
    
    # Add loading limit lines
    for row, col in [(1, 1), (1, 2), (2, 2)]:
        fig.add_hline(y=100, line_dash="dash", line_color="red",
                     annotation_text="100% Limit", row=row, col=col)
        fig.add_hline(y=90, line_dash="dash", line_color="orange",
                     annotation_text="90% Warning", row=row, col=col)
    
    fig.update_layout(
        title=f"Loading Analysis - {len(valid_branches)} Transmission Lines<br>" +
              f"Avg: {avg_loading:.1f}%, Max: {max_loading:.1f}%, Overloaded: {overloaded_count}",
        height=800,
        showlegend=False
    )
    
    return fig, {
        'avg_loading': avg_loading,
        'max_loading': max_loading,
        'overloaded_count': overloaded_count,
        'high_loading_count': high_loading_count,
        'total_branches': len(valid_branches)
    }

# Create loading analysis
if 'branches_df' in locals() and not branches_df.empty:
    loading_fig, loading_stats = create_loading_analysis(branches_df)
    loading_fig.show()
    
    print("📊 **Loading Analysis Results:**")
    print("=" * 40)
    print(f"📈 Average Loading: {loading_stats['avg_loading']:.1f}%")
    print(f"🔥 Maximum Loading: {loading_stats['max_loading']:.1f}%")
    print(f"❌ Overloaded Lines: {loading_stats['overloaded_count']} (>100%)")
    print(f"⚠️  High Loading Lines: {loading_stats['high_loading_count']} (>90%)")
    print(f"🔌 Total Branches: {loading_stats['total_branches']}")
else:
    print("❌ No branch data available for loading analysis")
````

## Database Analysis Functions

Comprehensive analysis functions for overall system performance and case-specific studies.

````python
# Database Analysis Functions
def get_overall_system_analysis(conn):
    """Perform comprehensive overall system analysis"""
    try:
        analysis_result = {
            'summary': 'Overall System Analysis',
            'timestamp': datetime.now().strftime('%Y-%m-%d %H:%M:%S'),
            'base_cases': [],
            'performance_metrics': {}
        }
        
        # Get all base cases
        base_cases_query = "SELECT DISTINCT base_case_id FROM BaseBusData ORDER BY base_case_id"
        base_cases = pd.read_sql_query(base_cases_query, conn)
        analysis_result['base_cases'] = base_cases['base_case_id'].tolist()
        
        # Overall voltage statistics
        voltage_stats_query = """
        SELECT 
            COUNT(*) as total_buses,
            AVG(VM) as avg_voltage,
            MIN(VM) as min_voltage,
            MAX(VM) as max_voltage,
            COUNT(CASE WHEN VM < 0.95 THEN 1 END) as low_voltage_count,
            COUNT(CASE WHEN VM > 1.05 THEN 1 END) as high_voltage_count
        FROM BaseBusData WHERE base_case_id = 0
        """
        voltage_stats = pd.read_sql_query(voltage_stats_query, conn)
        analysis_result['performance_metrics']['voltage'] = voltage_stats.iloc[0].to_dict()
        
        # Overall loading statistics
        loading_stats_query = """
        SELECT 
            COUNT(*) as total_branches,
            AVG(MVA/RATE * 100) as avg_loading,
            MAX(MVA/RATE * 100) as max_loading,
            COUNT(CASE WHEN MVA/RATE > 1.0 THEN 1 END) as overloaded_count
        FROM BaseBranchData WHERE base_case_id = 0 AND RATE > 0
        """
        loading_stats = pd.read_sql_query(loading_stats_query, conn)
        analysis_result['performance_metrics']['loading'] = loading_stats.iloc[0].to_dict()
        
        # SLR vs DLR comparison statistics
        slr_dlr_query = """
        SELECT 
            COUNT(*) as comparison_cases,
            AVG(s.VIO) as avg_slr_violation,
            AVG(d.VIO) as avg_dlr_violation,
            AVG((d.RATE - s.RATE)/s.RATE * 100) as avg_capacity_gain
        FROM SLR_Branches s 
        JOIN DLR_Branches d ON s.base_case_id = d.base_case_id 
            AND s.contingency_case_id = d.contingency_case_id
            AND s.From_Bus = d.From_Bus AND s.To_Bus = d.To_Bus
        WHERE s.base_case_id = 42 AND s.contingency_case_id = 123
        """
        slr_dlr_stats = pd.read_sql_query(slr_dlr_query, conn)
        if not slr_dlr_stats.empty:
            analysis_result['performance_metrics']['slr_dlr'] = slr_dlr_stats.iloc[0].to_dict()
        
        return analysis_result
        
    except Exception as e:
        return {'error': f'Overall analysis error: {str(e)}'}

def get_case_specific_analysis(conn, case_id=0, contingency_id=None):
    """Perform detailed case-specific analysis"""
    try:
        analysis_result = {
            'summary': f'Case-Specific Analysis - Base Case {case_id}',
            'case_id': case_id,
            'contingency_id': contingency_id,
            'timestamp': datetime.now().strftime('%Y-%m-%d %H:%M:%S'),
            'bus_analysis': {},
            'branch_analysis': {},
            'critical_elements': []
        }
        
        # Detailed bus analysis
        bus_query = f"""
        SELECT BUS_NUMBER, VM, VA, BASE_KV, PG, QG, PD, QD,
               CASE WHEN VM < 0.95 THEN 'Low Voltage'
                    WHEN VM > 1.05 THEN 'High Voltage'
                    ELSE 'Normal' END as voltage_status
        FROM BaseBusData WHERE base_case_id = {case_id}
        ORDER BY BUS_NUMBER
        """
        bus_data = pd.read_sql_query(bus_query, conn)
        
        # Detailed branch analysis
        branch_query = f"""
        SELECT branch_number, From_Bus, To_Bus, PF, QF, MVA, RATE, VIO,
               (MVA/RATE * 100) as loading_percent,
               CASE WHEN MVA/RATE > 1.0 THEN 'Overloaded'
                    WHEN MVA/RATE > 0.9 THEN 'Highly Loaded'
                    ELSE 'Normal' END as loading_status
        FROM BaseBranchData WHERE base_case_id = {case_id} AND RATE > 0
        ORDER BY loading_percent DESC
        """
        branch_data = pd.read_sql_query(branch_query, conn)
        
        # Find critical elements
        critical_buses = bus_data[bus_data['voltage_status'] != 'Normal']
        critical_branches = branch_data[branch_data['loading_status'] == 'Overloaded']
        
        analysis_result['bus_analysis'] = {
            'total_buses': len(bus_data),
            'voltage_violations': len(critical_buses),
            'critical_buses': critical_buses[['BUS_NUMBER', 'VM', 'voltage_status']].head(10).to_dict('records')
        }
        
        analysis_result['branch_analysis'] = {
            'total_branches': len(branch_data),
            'overloaded_branches': len(critical_branches),
            'critical_branches': critical_branches[['branch_number', 'From_Bus', 'To_Bus', 'loading_percent', 'loading_status']].head(10).to_dict('records')
        }
        
        return analysis_result
        
    except Exception as e:
        return {'error': f'Case analysis error: {str(e)}'}

def display_analysis_results(analysis_result):
    """Display analysis results in a formatted way"""
    if 'error' in analysis_result:
        print(f"❌ {analysis_result['error']}")
        return
    
    print(f"📊 **{analysis_result['summary']}**")
    print(f"🕒 **Generated:** {analysis_result['timestamp']}")
    print("=" * 60)
    
    if 'performance_metrics' in analysis_result:
        # Overall analysis display
        voltage_metrics = analysis_result['performance_metrics'].get('voltage', {})
        loading_metrics = analysis_result['performance_metrics'].get('loading', {})
        
        print("⚡ **Voltage Performance:**")
        print(f"   • Total Buses: {voltage_metrics.get('total_buses', 'N/A')}")
        print(f"   • Average Voltage: {voltage_metrics.get('avg_voltage', 0):.3f} p.u.")
        print(f"   • Voltage Range: {voltage_metrics.get('min_voltage', 0):.3f} - {voltage_metrics.get('max_voltage', 0):.3f} p.u.")
        print(f"   • Low Voltage Buses: {voltage_metrics.get('low_voltage_count', 0)}")
        print(f"   • High Voltage Buses: {voltage_metrics.get('high_voltage_count', 0)}")
        print()
        
        print("📈 **Loading Performance:**")
        print(f"   • Total Branches: {loading_metrics.get('total_branches', 'N/A')}")
        print(f"   • Average Loading: {loading_metrics.get('avg_loading', 0):.1f}%")
        print(f"   • Maximum Loading: {loading_metrics.get('max_loading', 0):.1f}%")
        print(f"   • Overloaded Lines: {loading_metrics.get('overloaded_count', 0)}")
        print()
        
        if 'slr_dlr' in analysis_result['performance_metrics']:
            slr_dlr_metrics = analysis_result['performance_metrics']['slr_dlr']
            print("🔄 **SLR vs DLR Analysis:**")
            print(f"   • Comparison Cases: {slr_dlr_metrics.get('comparison_cases', 'N/A')}")
            print(f"   • Average SLR Violation: {slr_dlr_metrics.get('avg_slr_violation', 0):.1f}%")
            print(f"   • Average DLR Violation: {slr_dlr_metrics.get('avg_dlr_violation', 0):.1f}%")
            print(f"   • Average Capacity Gain: {slr_dlr_metrics.get('avg_capacity_gain', 0):.1f}%")
    
    elif 'bus_analysis' in analysis_result:
        # Case-specific analysis display
        print(f"🎯 **Case Details:**")
        print(f"   • Base Case ID: {analysis_result['case_id']}")
        print(f"   • Contingency ID: {analysis_result.get('contingency_id', 'N/A')}")
        print()
        
        bus_analysis = analysis_result['bus_analysis']
        print("🚌 **Bus Analysis:**")
        print(f"   • Total Buses: {bus_analysis.get('total_buses', 0)}")
        print(f"   • Voltage Violations: {bus_analysis.get('voltage_violations', 0)}")
        print()
        
        branch_analysis = analysis_result['branch_analysis']
        print("⚡ **Branch Analysis:**")
        print(f"   • Total Branches: {branch_analysis.get('total_branches', 0)}")
        print(f"   • Overloaded Branches: {branch_analysis.get('overloaded_branches', 0)}")
        print()
        
        # Show critical elements
        if bus_analysis.get('critical_buses'):
            print("⚠️  **Critical Buses:**")
            for bus in bus_analysis['critical_buses'][:5]:
                print(f"   • Bus {bus['BUS_NUMBER']}: {bus['VM']:.3f} p.u. ({bus['voltage_status']})")
            print()
        
        if branch_analysis.get('critical_branches'):
            print("🔥 **Critical Branches:**")
            for branch in branch_analysis['critical_branches'][:5]:
                print(f"   • Line {branch['From_Bus']}-{branch['To_Bus']}: {branch['loading_percent']:.1f}% ({branch['loading_status']})")

print("✅ Database analysis functions ready!")
print("💡 Use: get_overall_system_analysis(conn) or get_case_specific_analysis(conn, case_id)")
````

## SLR vs DLR Comparison Analysis

Let's analyze the differences between Static Line Ratings (SLR) and Dynamic Line Ratings (DLR).

````python
# SLR vs DLR Comparison Analysis
def create_slr_dlr_comparison(comparison_df):
    """Create comprehensive SLR vs DLR comparison visualization"""
    
    if comparison_df.empty:
        fig = go.Figure()
        fig.add_annotation(
            text="No SLR/DLR comparison data available<br>(Requires contingency case data)",
            xref="paper", yref="paper", x=0.5, y=0.5, showarrow=False,
            font=dict(size=16)
        )
        fig.update_layout(title="SLR vs DLR Comparison - No Data Available", height=400)
        return fig, {}
    
    # Calculate metrics
    avg_slr_violation = comparison_df['SLR_VIO'].mean()
    avg_dlr_violation = comparison_df['DLR_VIO'].mean()
    capacity_gain = ((comparison_df['DLR_RATE'] - comparison_df['SLR_RATE']) / comparison_df['SLR_RATE'] * 100).mean()
    
    # Create subplots
    fig = make_subplots(
        rows=2, cols=2,
        subplot_titles=('Violation Comparison', 'Rating Comparison', 
                       'Capacity Gain Distribution', 'Efficiency Analysis'),
        specs=[[{"type": "scatter"}, {"type": "bar"}],
               [{"type": "histogram"}, {"type": "scatter"}]]
    )
    
    # Violation comparison
    fig.add_trace(
        go.Scatter(x=comparison_df['SLR_VIO'], y=comparison_df['DLR_VIO'],
                  mode='markers', name='SLR vs DLR Violations',
                  marker=dict(size=8, color='blue', opacity=0.6)),
        row=1, col=1
    )
    fig.add_trace(
        go.Scatter(x=[0, 100], y=[0, 100], mode='lines', 
                  name='Equal Line', line=dict(dash='dash', color='red')),
        row=1, col=1
    )
    
    # Rating comparison (sample branches)
    sample_branches = comparison_df.head(10)
    x_labels = [f"{int(row['From_Bus'])}-{int(row['To_Bus'])}" for _, row in sample_branches.iterrows()]
    
    fig.add_trace(
        go.Bar(x=x_labels, y=sample_branches['SLR_RATE'], 
              name='SLR Rating', marker_color='red', opacity=0.7),
        row=1, col=2
    )
    fig.add_trace(
        go.Bar(x=x_labels, y=sample_branches['DLR_RATE'], 
              name='DLR Rating', marker_color='blue', opacity=0.7),
        row=1, col=2
    )
    
    # Capacity gain distribution
    capacity_gains = (comparison_df['DLR_RATE'] - comparison_df['SLR_RATE']) / comparison_df['SLR_RATE'] * 100
    fig.add_trace(
        go.Histogram(x=capacity_gains, nbinsx=20, name='Capacity Gain Distribution',
                    marker_color='green', opacity=0.7),
        row=2, col=1
    )
    
    # Efficiency analysis
    efficiency_metric = comparison_df['DLR_VIO'] / comparison_df['SLR_VIO'].replace(0, 1)
    fig.add_trace(
        go.Scatter(x=comparison_df.index, y=efficiency_metric,
                  mode='markers', name='DLR Efficiency',
                  marker=dict(size=6, color=efficiency_metric, 
                             colorscale='RdYlGn', showscale=True)),
        row=2, col=2
    )
    
    fig.update_layout(
        title=f"SLR vs DLR Analysis - {len(comparison_df)} Comparison Cases<br>" +
              f"Avg Capacity Gain: {capacity_gain:.1f}%",
        height=800,
        showlegend=True
    )
    
    return fig, {
        'avg_slr_violation': avg_slr_violation,
        'avg_dlr_violation': avg_dlr_violation,
        'capacity_gain': capacity_gain,
        'comparison_cases': len(comparison_df)
    }

# Create SLR vs DLR comparison
if 'comparison_df' in locals():
    slr_dlr_fig, slr_dlr_stats = create_slr_dlr_comparison(comparison_df)
    slr_dlr_fig.show()
    
    if slr_dlr_stats:
        print("🔄 **SLR vs DLR Analysis Results:**")
        print("=" * 40)
        print(f"📊 Comparison Cases: {slr_dlr_stats['comparison_cases']}")
        print(f"📉 Average SLR Violation: {slr_dlr_stats['avg_slr_violation']:.1f}%")
        print(f"📈 Average DLR Violation: {slr_dlr_stats['avg_dlr_violation']:.1f}%")
        print(f"🚀 Average Capacity Gain: {slr_dlr_stats['capacity_gain']:.1f}%")
        print()
        print("💡 **Insights:**")
        print("   • DLR typically allows higher capacity utilization")
        print("   • Weather conditions significantly impact line ratings")
        print("   • Dynamic ratings can reduce violations under favorable conditions")
    else:
        print("ℹ️  No comparison data available - requires contingency case analysis")
else:
    print("❌ No comparison data available")
````

## Interactive AI Assistant

This section demonstrates AI-powered analysis capabilities using natural language commands.

````python
# AI Assistant Functions
def process_natural_language_query(query, conn):
    """Process natural language queries for power system analysis"""
    
    query_lower = query.lower()
    
    # Overall analysis requests
    if any(keyword in query_lower for keyword in ['overall analysis', 'system analysis', 'full analysis']):
        print("🔍 **Processing Overall Analysis Request...**")
        result = get_overall_system_analysis(conn)
        display_analysis_results(result)
        return result
    
    # Case-specific analysis requests
    elif any(keyword in query_lower for keyword in ['case analysis', 'analyze case', 'case study']):
        # Extract case ID from query
        import re
        numbers = re.findall(r'\d+', query)
        case_id = int(numbers[0]) if numbers else 0
        
        print(f"🎯 **Processing Case Analysis Request for Case {case_id}...**")
        result = get_case_specific_analysis(conn, case_id)
        display_analysis_results(result)
        return result
    
    # Database information requests
    elif any(keyword in query_lower for keyword in ['database info', 'table info', 'list tables']):
        print("📋 **Database Information:**")
        get_database_info(conn)
        return {'type': 'database_info'}
    
    # Available cases requests
    elif any(keyword in query_lower for keyword in ['list cases', 'available cases', 'show cases']):
        try:
            cases_query = "SELECT DISTINCT base_case_id FROM BaseBusData ORDER BY base_case_id"
            cases_df = pd.read_sql_query(cases_query, conn)
            
            contingency_query = "SELECT DISTINCT base_case_id, contingency_case_id FROM SLR_Branches ORDER BY base_case_id, contingency_case_id"
            contingency_df = pd.read_sql_query(contingency_query, conn)
            
            print("📋 **Available Cases:**")
            print(f"🏠 **Base Cases:** {', '.join(map(str, cases_df['base_case_id'].tolist()))}")
            print(f"🔄 **Contingency Cases:** {len(contingency_df)} available")
            
            return {'type': 'cases_list', 'base_cases': cases_df['base_case_id'].tolist()}
        except Exception as e:
            print(f"❌ Error retrieving cases: {e}")
            return {'error': str(e)}
    
    # Visualization requests
    elif any(keyword in query_lower for keyword in ['voltage analysis', 'show voltage', 'voltage plot']):
        print("⚡ **Generating Voltage Analysis...**")
        if 'buses_df' in globals() and not buses_df.empty:
            voltage_fig, voltage_stats = create_voltage_analysis(buses_df)
            voltage_fig.show()
            return {'type': 'visualization', 'data': voltage_stats}
        else:
            print("❌ No bus data available for voltage analysis")
            return {'error': 'No bus data available'}
    
    elif any(keyword in query_lower for keyword in ['loading analysis', 'show loading', 'loading plot']):
        print("📊 **Generating Loading Analysis...**")
        if 'branches_df' in globals() and not branches_df.empty:
            loading_fig, loading_stats = create_loading_analysis(branches_df)
            loading_fig.show()
            return {'type': 'visualization', 'data': loading_stats}
        else:
            print("❌ No branch data available for loading analysis")
            return {'error': 'No branch data available'}
    
    elif any(keyword in query_lower for keyword in ['slr vs dlr', 'comparison', 'dlr comparison']):
        print("🔄 **Generating SLR vs DLR Comparison...**")
        if 'comparison_df' in globals():
            slr_dlr_fig, slr_dlr_stats = create_slr_dlr_comparison(comparison_df)
            slr_dlr_fig.show()
            return {'type': 'visualization', 'data': slr_dlr_stats}
        else:
            print("❌ No comparison data available")
            return {'error': 'No comparison data available'}
    
    else:
        print(f"🤔 **Query:** {query}")
        print("💡 **Available Commands:**")
        print("   • 'overall analysis' - Complete system analysis")
        print("   • 'case analysis [number]' - Specific case study")
        print("   • 'list cases' - Show available cases")
        print("   • 'voltage analysis' - Voltage visualization")
        print("   • 'loading analysis' - Loading visualization") 
        print("   • 'slr vs dlr' - Comparison analysis")
        print("   • 'database info' - Database structure")
        return {'type': 'help'}

# AI Assistant Interface
def ai_assistant_interface():
    """Interactive AI assistant for power system analysis"""
    print("🤖 **AI Power System Assistant**")
    print("=" * 50)
    print("I can help you with comprehensive power system analysis!")
    print()
    print("💡 **Try these commands:**")
    print("   • 'overall analysis'")
    print("   • 'case analysis 0'") 
    print("   • 'list cases'")
    print("   • 'voltage analysis'")
    print("   • 'loading analysis'")
    print("   • 'slr vs dlr'")
    print()
    
    while True:
        try:
            user_query = input("🎯 Enter your query (or 'quit' to exit): ").strip()
            
            if user_query.lower() in ['quit', 'exit', 'q']:
                print("👋 Thank you for using the AI Power System Assistant!")
                break
            
            if user_query:
                print(f"\n🔍 **Processing:** {user_query}")
                print("-" * 50)
                result = process_natural_language_query(user_query, conn)
                print("\n" + "="*50)
            else:
                print("ℹ️  Please enter a valid query.")
                
        except KeyboardInterrupt:
            print("\n👋 Assistant session ended.")
            break
        except Exception as e:
            print(f"❌ Error: {e}")

print("✅ AI Assistant functions ready!")
print("💡 Use: ai_assistant_interface() for interactive mode")
print("💡 Use: process_natural_language_query('your query', conn) for direct queries")
````
